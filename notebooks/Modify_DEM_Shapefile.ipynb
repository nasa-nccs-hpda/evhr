# Modify DEM Shapefile

EVHR requires a DEM shapefile to point at the files directly related to the DEM processing.
The following notebook takes care of modifying the existing shapefile provided in the EVHR
repository to mimic new paths in the event EVHR is ran in a different compute system
other than Explore.

The current shapefile is available at [evrh/model/SRTM](https://github.com/nasa-nccs-hpda/evhr/tree/main/model/SRTM).
The output from this notebook is a new shapefile with the locations modified
to match the new path.

```python
import geopandas as gpd
```

```python
input_dem_shapefile = '../model/SRTM/srtm.shp'
existing_dem_path = '/explore/nobackup/projects/dem/SRTM/1-ArcSec'
new_dem_path = '/discover/nobackup/projects/akmosaic/dem/SRTM/1-ArcSec'
output_dem_shapefile = '../model/SRTM/srtm_discover.shp'
```

```python
gdf = gpd.read_file(input_dem_shapefile)
gdf.shape
```

```python
# example row
gdf.iloc[0]
```

```python
gdf['location'] = gdf['location'].str.replace(
    existing_dem_path,
    new_dem_path
)
```

```python
gdf.head()
```

```python
gdf.to_file(output_dem_shapefile)
```
